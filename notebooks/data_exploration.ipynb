```python
import pandas as pd
import numpy as np

df_udp = pd.read_csv('../data/raw/DrDoS_UDP.csv', nrows=200000)  
print(f"UDP loaded: {df_udp.shape}")

df_syn = pd.read_csv('../data/raw/Syn.csv', nrows=200000)
print(f'SYN Loaded: {df_syn.shape}')
```

```python
df_udp.columns = df_udp.columns.str.strip()
df_syn.columns = df_syn.columns.str.strip()

print("UDP labels:", df_udp['Label'].unique())
print("SYN labels:", df_syn['Label'].unique())

print("\nUDP label counts:")
print(df_udp['Label'].value_counts())

print("\nSYN label counts:")
print(df_syn['Label'].value_counts())
```

```python
df_udp['Attack_Type'] = df_udp['Label'].apply(
    lambda x: 'UDP_Flood' if x != 'BENIGN' else 'BENIGN'
)
df_syn['Attack_Type'] = df_syn['Label'].apply(
    lambda x: 'SYN_Flood' if x != 'BENIGN' else 'BENIGN'
)

df = pd.concat([df_udp, df_syn], ignore_index=True)
print(f"Combined dataset shape: {df.shape}")
print("\nAttack type distribution:")
print(df['Attack_Type'].value_counts())
```

```python
selected_features = [
    'Flow Duration',
    'Total Fwd Packets',
    'Total Backward Packets',
    'Total Length of Fwd Packets',
    'Total Length of Bwd Packets',
    'Flow Bytes/s',
    'Flow Packets/s',
    'Fwd Packet Length Mean',
    'Bwd Packet Length Mean',
    'Fwd Packet Length Std',
    'Bwd Packet Length Std',
    'Fwd IAT Mean',        
    'Bwd IAT Mean',
    'Packet Length Mean',
    'Packet Length Std',
    'Label',
    'Attack_Type'
]

available = [col for col in selected_features 
             if col in df.columns]
missing = [col for col in selected_features 
           if col not in df.columns]

print("Available features:", available)
print("\nMissing features:", missing)

df_selected = df[available].copy()
print(f"\nDataset with selected features: {df_selected.shape}")
```

```python
df_selected.replace([np.inf, -np.inf], np.nan, inplace=True)

print("Missing values before cleaning:")
print(df_selected.isnull().sum())

df_selected.dropna(inplace=True)

print(f"\nShape after cleaning: {df_selected.shape}")
print("\nMissing values after cleaning:")
print(df_selected.isnull().sum().sum(), "total missing values")
```

```python
df_selected['Binary_Label'] = df_selected['Label'].apply(
    lambda x: 0 if x == 'BENIGN' else 1
)

print("Label distribution before balancing:")
print(df_selected['Binary_Label'].value_counts())
print(f"Total rows: {len(df_selected)}")

normal = df_selected[df_selected['Binary_Label'] == 0]
attack = df_selected[df_selected['Binary_Label'] == 1]

print(f"\nNormal samples: {len(normal)}")
print(f"Attack samples: {len(attack)}")

min_samples = min(len(normal), len(attack), 50000)

normal_sample = normal.sample(n=min_samples, random_state=42)
attack_sample = attack.sample(n=min_samples, random_state=42)

df_balanced = pd.concat([normal_sample, attack_sample], 
                         ignore_index=True)

df_balanced = df_balanced.sample(frac=1, 
                                  random_state=42).reset_index(drop=True)

print(f"\nBalanced dataset shape: {df_balanced.shape}")
print("Label distribution after balancing:")
print(df_balanced['Binary_Label'].value_counts())
```

```python
df_balanced.to_csv('../data/processed/clean_dataset.csv', index=False)
print("Saved to data/processed/clean_dataset.csv")

print(f"\nFinal dataset summary:")
print(f"  Total rows: {len(df_balanced)}")
print(f"  Features: {len(df_balanced.columns) - 3} input features")
print(f"  Normal traffic: {len(df_balanced[df_balanced['Binary_Label']==0])}")
print(f"  Attack traffic: {len(df_balanced[df_balanced['Binary_Label']==1])}")
print(f"  Attack types: {df_balanced['Attack_Type'].unique()}")
print("\nPhase 2 complete. Ready for EDA.")
```
